# Day 3 – Functions, Lambda and File Handling

## Functions

```python
def greet(name):
    return f'Hello {name}! Welcome to Data Science.'

print(greet('Ganesh'))
```

## Function with multiple arguments

```python
def calculate_average(a, b, c):
    return (a + b + c) / 3

print(calculate_average(80, 90, 70))
```

## Lambda

```python
square = lambda x: x ** 2
print(square(8))
```

## List comprehension

```python
squares = [x ** 2 for x in range(1, 11)]
print(squares)
```

## File Handling – Write

```python
data = 'Python\nSQL\nPower BI\nMachine Learning'

with open('skills.txt', 'w', encoding='utf-8') as file:
    file.write(data)

print('skills.txt created successfully.')
```

## File Handling – Read

```python
with open('skills.txt', 'r', encoding='utf-8') as file:
    content = file.read()

print(content)
```

## Exception Handling

```python
try:
    a = float(input('Enter numerator: '))
    b = float(input('Enter denominator: '))
    print('Result:', a / b)
except ZeroDivisionError:
    print('Denominator cannot be zero.')
except ValueError:
    print('Please enter valid numbers.')
finally:
    print('Program completed.')
```

## Mini Task – Reusable Student Result Function

```python
def student_result(name, marks):
    total = sum(marks)
    average = total / len(marks)
    status = 'Pass' if average >= 40 else 'Fail'
    return {'name': name, 'total': total, 'average': average, 'status': status}

result = student_result('Ganesh', [78, 85, 91, 82, 88])
print(result)
```
