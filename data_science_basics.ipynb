# Day 4 – Data Science Fundamentals

## What is Data Science?

Data Science combines programming, statistics, mathematics, data analysis and machine learning to extract useful insights from data and support decisions.

## Common Data Science Workflow

1. Define the problem
2. Collect data
3. Clean data
4. Explore data (EDA)
5. Visualize data
6. Build a model when needed
7. Evaluate results
8. Communicate insights

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## NumPy Basics

```python
arr = np.array([10, 20, 30, 40, 50])
print('Array:', arr)
print('Mean:', np.mean(arr))
print('Maximum:', np.max(arr))
print('Minimum:', np.min(arr))
print('Standard deviation:', np.std(arr))
```

## Pandas Basics

```python
data = {
    'Name': ['Ganesh', 'Ravi', 'Sita', 'Anu', 'Kiran'],
    'Python': [82, 74, 91, 68, 88],
    'SQL': [78, 81, 89, 72, 84],
    'PowerBI': [85, 76, 93, 70, 90]
}

df = pd.DataFrame(data)
df['Average'] = df[['Python', 'SQL', 'PowerBI']].mean(axis=1)
print(df)
```

```python
print('Shape:', df.shape)
print('\nSummary:')
print(df.describe())
print('\nTop performers:')
print(df.sort_values('Average', ascending=False).head(3))
```

## Simple Visualization

```python
plt.figure(figsize=(8, 4))
plt.bar(df['Name'], df['Average'])
plt.title('Student Average Scores')
plt.xlabel('Student')
plt.ylabel('Average Score')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
