# Day 2 – Control Flow, Lists and Dictionaries

## if / elif / else

```python
marks = float(input('Enter marks: '))

if marks >= 90:
    grade = 'A'
elif marks >= 75:
    grade = 'B'
elif marks >= 60:
    grade = 'C'
elif marks >= 40:
    grade = 'D'
else:
    grade = 'F'

print('Grade:', grade)
```

## for loop

```python
for i in range(1, 11):
    print(i)
```

## while loop

```python
n = 5
while n > 0:
    print(n)
    n -= 1
```

## Lists

```python
marks = [78, 85, 92, 67, 88]
print('Marks:', marks)
print('First mark:', marks[0])
print('Highest:', max(marks))
print('Lowest:', min(marks))
print('Average:', sum(marks) / len(marks))

marks.append(95)
marks.remove(67)
print('Updated:', marks)
```

## Tuples and Sets

```python
subjects = ('Python', 'SQL', 'Power BI')
print('Tuple:', subjects)

skills = {'Python', 'SQL', 'Python', 'Power BI'}
print('Set removes duplicates:', skills)
```

## Dictionaries

```python
student = {
    'name': 'Ganesh',
    'age': 23,
    'course': 'Data Science',
    'score': 82
}

print(student['name'])
print(student.get('course'))
student['score'] = 88
student['city'] = 'Hyderabad'
print(student)
```

## Practice – Pass/Fail for Multiple Students

```python
students = {'Ganesh': 82, 'Ravi': 36, 'Sita': 91, 'Anu': 67}

for name, score in students.items():
    result = 'Pass' if score >= 40 else 'Fail'
    print(name, score, result)
```

## Challenge – Count Even and Odd Numbers

```python
numbers = list(range(1, 21))
even = 0
odd = 0

for n in numbers:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1

print('Even count:', even)
print('Odd count:', odd)
```
