# Day 1 – Python Basics

This notebook covers Python installation/setup concepts, variables, data types, input/output, type conversion, operators, and beginner practice.

```python
# First Python program
print('Hello, Data Science!')
```

## Variables and Data Types

```python
name = 'Ganesh'
age = 23
percentage = 72.5
is_student = True

print(name, type(name))
print(age, type(age))
print(percentage, type(percentage))
print(is_student, type(is_student))
```

## Input and Type Conversion

```python
name = input('Enter your name: ')
age = int(input('Enter your age: '))

print(f'Hello {name}, you are {age} years old.')
```

## Operators

```python
a = 20
b = 6

print('Addition:', a + b)
print('Subtraction:', a - b)
print('Multiplication:', a * b)
print('Division:', a / b)
print('Floor Division:', a // b)
print('Modulus:', a % b)
print('Power:', a ** b)

print('20 > 6:', a > b)
print('20 == 6:', a == b)
print('(a > 10 and b < 10):', a > 10 and b < 10)
```

## Practice 1 – Simple Calculator

```python
num1 = float(input('Enter first number: '))
num2 = float(input('Enter second number: '))

print('Sum =', num1 + num2)
print('Difference =', num1 - num2)
print('Product =', num1 * num2)
print('Average =', (num1 + num2) / 2)
```

## Practice 2 – Student Result

```python
name = input('Student name: ')
maths = float(input('Maths marks: '))
python = float(input('Python marks: '))
sql = float(input('SQL marks: '))

total = maths + python + sql
average = total / 3

print('\nStudent:', name)
print('Total:', total)
print('Average:', average)
```

## Practice 3 – Temperature Conversion

```python
celsius = float(input('Temperature in Celsius: '))
fahrenheit = (celsius * 9/5) + 32
print('Fahrenheit:', fahrenheit)
```

## Day 1 Challenge – Salary Calculator

```python
basic_salary = float(input('Enter basic salary: '))
hra = basic_salary * 0.20
da = basic_salary * 0.10
final_salary = basic_salary + hra + da
print('HRA:', hra)
print('DA:', da)
print('Final salary:', final_salary)
```
