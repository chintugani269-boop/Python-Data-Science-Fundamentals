# Mini Project – Student Performance Analysis using Python

## Objective
Analyze student academic performance using Python, NumPy, Pandas and Matplotlib, and identify high/low performers and subject-wise patterns.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Create the dataset

```python
data = {
    'Student_ID': range(1, 11),
    'Name': ['Ganesh', 'Ravi', 'Sita', 'Anu', 'Kiran', 'Priya', 'Rahul', 'Sneha', 'Arjun', 'Divya'],
    'Python': [82, 74, 91, 68, 88, 79, 65, 94, 72, 86],
    'SQL': [78, 81, 89, 72, 84, 76, 70, 92, 75, 83],
    'PowerBI': [85, 76, 93, 70, 90, 81, 68, 95, 78, 88],
    'Attendance': [92, 85, 96, 78, 90, 88, 72, 98, 80, 91]
}

df = pd.DataFrame(data)
df.head()
```

## 2. Calculate average and result

```python
df['Average_Marks'] = df[['Python', 'SQL', 'PowerBI']].mean(axis=1).round(2)
df['Result'] = np.where(df['Average_Marks'] >= 40, 'Pass', 'Fail')
df
```

## 3. Basic analysis

```python
print('Total students:', len(df))
print('Class average:', round(df['Average_Marks'].mean(), 2))
print('Highest average:', df['Average_Marks'].max())
print('Lowest average:', df['Average_Marks'].min())
print('Pass count:', (df['Result'] == 'Pass').sum())
print('Fail count:', (df['Result'] == 'Fail').sum())
```

## 4. Top performers

```python
top_students = df.sort_values('Average_Marks', ascending=False).head(3)
print(top_students[['Name', 'Average_Marks', 'Attendance']])
```

## 5. Subject-wise averages

```python
subject_averages = df[['Python', 'SQL', 'PowerBI']].mean().round(2)
print(subject_averages)
```

## 6. Attendance analysis

```python
df['Attendance_Category'] = np.where(df['Attendance'] >= 85, 'Good', 'Needs Improvement')
print(df[['Name', 'Attendance', 'Attendance_Category']])
```

## 7. Visualizations

```python
plt.figure(figsize=(9, 4))
plt.bar(df['Name'], df['Average_Marks'])
plt.title('Student Average Marks')
plt.xlabel('Student')
plt.ylabel('Average Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
subject_averages.plot(kind='bar')
plt.title('Subject-wise Average')
plt.xlabel('Subject')
plt.ylabel('Average Marks')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.scatter(df['Attendance'], df['Average_Marks'])
plt.title('Attendance vs Average Marks')
plt.xlabel('Attendance (%)')
plt.ylabel('Average Marks')
plt.tight_layout()
plt.show()
```

## 8. Key Insights

- The dataset contains 10 students.
- Average marks are calculated across Python, SQL and Power BI.
- Top performers can be identified by sorting Average_Marks.
- Subject-wise averages show which subject has the strongest overall performance.
- Attendance can be categorized to identify students who may need improvement.
- The scatter plot helps visually inspect the relationship between attendance and marks.
